### Initializate SDK Client

```python
from google import genai

client = genai.Client()
```

## Tokens in the Gemini API

```python
MODEL_ID = "gemini-2.5-flash"
```

```python
from google.genai import types

response = client.models.generate_content(
    model=MODEL_ID,
    contents="What's the highest mountain in Africa?",
    config=types.GenerateContentConfig(
        temperature=0.1,
        max_output_tokens=500,        
        candidate_count=3
    ),
)

print("===============================================")
print("Candidates:")
i = 0
for candidate in response.candidates:
    i += 1
    print(f"{i}º Answer:")
    print(candidate.token_count)
    print(f"Avg log probs: {candidate.avg_logprobs}")
    print(f"Answer candidate:")
    for part in candidate.content.parts:
        print(part.text)
    print("===============================================")

print(f"Final answer:\n {response.text}")
print("===============================================")

print("Tokens")
print(f"Thoughts tokens: {response.usage_metadata.thoughts_token_count}")
print(f"Total candidates token count: {response.usage_metadata.candidates_token_count}")
j = 0
for token_candidate in response.usage_metadata.candidates_tokens_details:
    j += 1
    print(f"\t {j}º candidates token count: {token_candidate.token_count}")
print(f"Output Token count: {response.usage_metadata.candidates_token_count + response.usage_metadata.thoughts_token_count}")
print(f"Input token count: {response.usage_metadata.prompt_token_count}")
print(f"Total token count: {response.usage_metadata.total_token_count}")
```
